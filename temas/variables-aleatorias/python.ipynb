NumPy permite generar muestras de diferentes distribuciones de probabilidad.

### 1.2 Estadísticos con NumPy

```python
# Estadísticos básicos
datos = np.random.normal(loc=50, scale=10, size=100)

media = np.mean(datos)
mediana = np.median(datos)
desviacion = np.std(datos)
varianza = np.var(datos)
minimo = np.min(datos)
maximo = np.max(datos)

print(f"Media: {media:.2f}")
print(f"Mediana: {mediana:.2f}")
print(f"Desviación estándar: {desviacion:.2f}")
print(f"Varianza: {varianza:.2f}")
print(f"Rango: [{minimo:.2f}, {maximo:.2f}]")
```

```python
# Percentiles y cuantiles
q25 = np.percentile(datos, 25)
q50 = np.percentile(datos, 50)  # mediana
q75 = np.percentile(datos, 75)
iqr = q75 - q25  # Rango intercuartílico

print(f"Q1 (25%): {q25:.2f}")
print(f"Q2 (50%): {q50:.2f}")
print(f"Q3 (75%): {q75:.2f}")
print(f"IQR: {iqr:.2f}")
```

## 2. Pandas: Análisis y Manipulación de Datos

Pandas proporciona estructuras de datos de alto nivel (Series y DataFrame) y herramientas para análisis de datos.

```python
# Series: array unidimensional con índice
serie = pd.Series([10, 20, 30, 40, 50], index=["a", "b", "c", "d", "e"])
print("Serie:")
print(serie)
print(f"\nAcceso por índice: serie['c'] = {serie['c']}")
```

```python
# DataFrame: tabla bidimensional con etiquetas
datos_clientes = {
    "nombre": ["Ana", "Pedro", "María", "Juan", "Laura"],
    "edad": [25, 32, 28, 45, 37],
    "ciudad": ["Madrid", "Barcelona", "Madrid", "Valencia", "Barcelona"],
    "satisfaccion": [5, 4, 5, 3, 4],
    "compras": [12, 8, 15, 5, 10],
}

df = pd.DataFrame(datos_clientes)
print("DataFrame:")
print(df)
```

### 2.1 Exploración de Datos

```python
# Información básica del DataFrame
print("Forma del DataFrame:", df.shape)
print("\nTipos de datos:")
print(df.dtypes)
print("\nInformación general:")
df.info()
```

```python
# Primeras y últimas filas
print("Primeras 3 filas:")
print(df.head(3))
print("\nÚltimas 2 filas:")
print(df.tail(2))
```

```python
# Estadísticos descriptivos
print("Descripción estadística:")
print(df.describe())
```

### 2.2 Selección y Filtrado

```python
# Seleccionar columnas
edades = df["edad"]
print("Columna edad:")
print(edades)

# Seleccionar múltiples columnas
subset = df[["nombre", "satisfaccion"]]
print("\nSubset de columnas:")
print(subset)
```

```python
# Filtrar filas por condición
clientes_madrid = df[df["ciudad"] == "Madrid"]
print("Clientes de Madrid:")
print(clientes_madrid)

# Filtros múltiples
clientes_satisfechos = df[(df["satisfaccion"] >= 4) & (df["compras"] > 8)]
print("\nClientes satisfechos con muchas compras:")
print(clientes_satisfechos)
```

### 2.3 Agregaciones y Agrupaciones

```python
# Estadísticos por columna
print("Edad media:", df["edad"].mean())
print("Compras totales:", df["compras"].sum())
print("Satisfacción máxima:", df["satisfaccion"].max())
```

```python
# Agrupar por categorías
por_ciudad = df.groupby("ciudad").agg(
    {"edad": "mean", "satisfaccion": "mean", "compras": "sum"}
)

print("Agregaciones por ciudad:")
print(por_ciudad)
```

### 2.4 Crear Nuevas Columnas

```python
# Operaciones aritméticas
df["compras_por_año_edad"] = df["compras"] / df["edad"]

# Transformaciones condicionales
df["cliente_premium"] = df["compras"].apply(lambda x: "Sí" if x >= 10 else "No")

# Categorización
df["grupo_edad"] = pd.cut(
    df["edad"], bins=[0, 30, 40, 100], labels=["Joven", "Adulto", "Senior"]
)

print(df)
```

## 3. Matplotlib: Visualización de Datos

Matplotlib es la librería fundamental para crear gráficos en Python.

### 3.1 Histogramas - Distribución Empírica

```python
# Generar datos de distribución normal
np.random.seed(42)
alturas = np.random.normal(loc=170, scale=10, size=1000)

plt.figure(figsize=(10, 6))
plt.hist(alturas, bins=30, density=True, alpha=0.7, color="skyblue", edgecolor="black")
plt.axvline(
    np.mean(alturas),
    color="red",
    linestyle="--",
    linewidth=2,
    label=f"Media: {np.mean(alturas):.1f}",
)
plt.axvline(
    np.median(alturas),
    color="green",
    linestyle="--",
    linewidth=2,
    label=f"Mediana: {np.median(alturas):.1f}",
)
plt.xlabel("Altura (cm)")
plt.ylabel("Densidad")
plt.title("Distribución de Alturas - Histograma")
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```

### 3.5 Múltiples Subgráficos

```python
# Comparar diferentes distribuciones
fig, axes = plt.subplots(2, 2, figsize=(14, 10))

# Distribución Uniforme
uniforme = np.random.uniform(0, 10, 1000)
axes[0, 0].hist(uniforme, bins=30, color="lightblue", edgecolor="black")
axes[0, 0].set_title("Uniforme(0, 10)")
axes[0, 0].set_xlabel("Valor")
axes[0, 0].set_ylabel("Frecuencia")

# Distribución Normal
normal = np.random.normal(0, 1, 1000)
axes[0, 1].hist(normal, bins=30, color="lightgreen", edgecolor="black")
axes[0, 1].set_title("Normal(0, 1)")
axes[0, 1].set_xlabel("Valor")
axes[0, 1].set_ylabel("Frecuencia")

# Distribución Binomial
binomial = np.random.binomial(10, 0.5, 1000)
axes[1, 0].hist(binomial, bins=11, color="salmon", edgecolor="black")
axes[1, 0].set_title("Binomial(n=10, p=0.5)")
axes[1, 0].set_xlabel("Número de éxitos")
axes[1, 0].set_ylabel("Frecuencia")

# Distribución Poisson
poisson = np.random.poisson(5, 1000)
axes[1, 1].hist(poisson, bins=range(0, 20), color="wheat", edgecolor="black")
axes[1, 1].set_title("Poisson(λ=5)")
axes[1, 1].set_xlabel("Número de eventos")
axes[1, 1].set_ylabel("Frecuencia")

# Establecer el mismo eje X en todos los subgráficos
global_min = min(uniforme.min(), normal.min(), binomial.min(), poisson.min())
global_max = max(uniforme.max(), normal.max(), binomial.max(), poisson.max())
for ax in axes.ravel():
    ax.set_xlim(global_min, global_max)

plt.tight_layout()
plt.show()
```

## 4. Scipy.stats: Distribuciones de Probabilidad

Scipy proporciona funciones para trabajar con distribuciones de probabilidad teóricas.

```python
# Distribución Normal
mu, sigma = 0, 1
x = np.linspace(-4, 4, 100)
pdf = stats.norm.pdf(x, mu, sigma)  # Función de densidad
cdf = stats.norm.cdf(x, mu, sigma)  # Función de distribución acumulada

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))

# PDF
ax1.plot(x, pdf, "b-", linewidth=2, label="PDF")
ax1.fill_between(x, pdf, alpha=0.3)
ax1.set_xlabel("x")
ax1.set_ylabel("Densidad")
ax1.set_title("Función de Densidad de Probabilidad (PDF)")
ax1.legend()
ax1.grid(True, alpha=0.3)

# CDF
ax2.plot(x, cdf, "r-", linewidth=2, label="CDF")
ax2.set_xlabel("x")
ax2.set_ylabel("Probabilidad acumulada")
ax2.set_title("Función de Distribución Acumulada (CDF)")
ax2.legend()
ax2.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

```python
# Calcular probabilidades con scipy
print("=== Probabilidades con Distribución Normal(0, 1) ===")
print(f"P(X <= 0) = {stats.norm.cdf(0, 0, 1):.4f}")
print(f"P(X <= 1.96) = {stats.norm.cdf(1.96, 0, 1):.4f}")
print(
    f"P(-1.96 <= X <= 1.96) = {stats.norm.cdf(1.96, 0, 1) - stats.norm.cdf(-1.96, 0, 1):.4f}"
)

# Calcular cuantiles (inverso de CDF)
print("\n=== Cuantiles ===")
print(f"Cuantil 0.025: {stats.norm.ppf(0.025, 0, 1):.4f}")
print(f"Cuantil 0.5 (mediana): {stats.norm.ppf(0.5, 0, 1):.4f}")
print(f"Cuantil 0.975: {stats.norm.ppf(0.975, 0, 1):.4f}")
```

### 4.1 Teorema del Límite Central - Demostración

```python
# Demostración del Teorema del Límite Central
np.random.seed(40)

# Distribución original no necesariamente normal: Uniforme en este caso
# Calcular medias muestrales con diferentes tamaños
"""
Para comparar correctamente las distribuciones de medias, mantenemos el mismo eje X
para todos los subgráficos. Primero precomputamos las medias y obtenemos el rango global.
Además, mostraremos la media y desviación típica teóricas (TLC) y las observadas
para cada distribución de medias muestrales.
"""


# Tamaños muestrales a evaluar
tamaños = [5, 10, 30, 100]
medias_por_n = []
for n in tamaños:
    medias_n = np.array([np.random.uniform(0, 10, n).mean() for _ in range(1000)])
    medias_por_n.append(medias_n)

# Rango global de todas las medias
global_min = min(m.min() for m in medias_por_n)
global_max = max(m.max() for m in medias_por_n)

fig, axes = plt.subplots(2, 2, figsize=(14, 10))
axes = axes.ravel()

# Parámetros teóricos (poblacionales)
mu = 5
sigma = np.std(np.arange(0, 11))

for i, (n, medias) in enumerate(zip(tamaños, medias_por_n)):
    # Histograma de las medias muestrales
    axes[i].hist(
        medias,
        bins=30,
        density=True,
        alpha=0.7,
        color="lightblue",
        edgecolor="black",
    )

    # Estadísticos observados de las medias simuladas
    mu_obs = medias.mean()
    sigma_obs = medias.std()  # ddof=0 para comparar con σ/√n

    # Curva normal teórica y observada del TLC con el mismo eje X
    sigma_teorica = sigma / np.sqrt(n)
    x = np.linspace(global_min, global_max, 300)
    axes[i].plot(
        x,
        stats.norm.pdf(x, mu, sigma_teorica),
        "r-",
        linewidth=2,
        label="Normal teórica",
    )
    axes[i].plot(
        x,
        stats.norm.pdf(x, mu_obs, sigma_obs),
        "b-",
        linewidth=2,
        label="Normal estimada",
    )

    # Resumen numérico en una cajita dentro del plot
    txt = (
        f"μ observada = {mu_obs:.3f}\n"
        f"μ teórica = {mu:.3f}\n"
        f"σ observada = {sigma_obs:.3f}\n"
        f"σ teórica = {sigma_teorica:.3f}"
    )
    axes[i].text(
        0.98,
        0.98,
        txt,
        transform=axes[i].transAxes,
        ha="right",
        va="top",
        fontsize=10,
        bbox=dict(boxstyle="round", facecolor="white", alpha=0.75, edgecolor="#555"),
    )

    axes[i].set_title(f"Distribución de medias muestrales (n={n})")
    axes[i].set_xlabel("Media muestral")
    axes[i].set_ylabel("Densidad")
    axes[i].legend(loc="upper left")
    axes[i].grid(True, alpha=0.3)
    axes[i].set_xlim(global_min, global_max)

plt.tight_layout()
plt.show()

print("Observación: A medida que n aumenta, la distribución de las medias muestrales")
print("se aproxima a una distribución normal y su dispersión disminuye como σ/√n.")
```
