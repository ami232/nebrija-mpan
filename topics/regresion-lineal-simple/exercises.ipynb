## Parte 5: Variable independiente ordinal - Nivel educativo y salario

### Ejercicio 5.1: Regresión con variable ordinal

Las variables ordinales tienen un orden natural pero no necesariamente intervalos iguales. Podemos codificarlas numéricamente respetando el orden.

Tenemos datos de salarios según el nivel educativo: Secundaria, Grado, Máster, Doctorado.

```python
# Generar datos sintéticos
np.random.seed(456)
niveles = ["Secundaria", "Grado", "Máster", "Doctorado"]
n_por_nivel = 15

educacion = np.repeat(niveles, n_por_nivel)
# Salarios crecientes por nivel educativo
salario_educacion = np.concatenate(
    [
        np.random.normal(25000, 2500, n_por_nivel),  # Secundaria
        np.random.normal(32000, 3000, n_por_nivel),  # Grado
        np.random.normal(47000, 4000, n_por_nivel),  # Máster
        np.random.normal(55000, 5000, n_por_nivel),  # Doctorado
    ]
)

datos_educacion = pd.DataFrame({"Educacion": educacion, "Salario": salario_educacion})

print(datos_educacion.head(10))
print(f"\nSalario medio por nivel:")
print(datos_educacion.groupby("Educacion")["Salario"].mean().round(0))
```
