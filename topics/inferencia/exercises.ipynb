El teorema del límite central dice que si:
- $X_1, X_2, \dots, X_n, \dots$ son muestras aleatorias extraídas de una población con media global $\mu$ y varianza finita $\sigma^2$
- $\bar{X}_n$ es la media muestral de las primeras $n$ muestras


entonces la forma límite de la distribución,

$$\lim_{n \to \infty} {\left ( \frac{\bar{X}_n-\mu}{\sigma_\bar{X}} \right ) = N(0, 1)}$$

Es decir, podemos aproximar $\bar{X}_n$ es una variable aleatoria cuya distribución converge a una distribución normal

$$N(\mu, \sigma/\sqrt{n})$$

Y podemos usarla para determinar qué valores son más o menos probables de $\bar{X}_n$.
