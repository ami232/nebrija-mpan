**Distribuciones disponibles:**
- `stats.norm` - Distribución normal

- `stats.binom` - Distribución binomial
- `stats.bernoulli` - Distribución de Bernoulli

**Funciones principales:**
- `.pmf(k, ...)` - Función de masa de probabilidad (distribuciones discretas)
- `.pdf(x, ...)` - Función de densidad de probabilidad (distribuciones continuas)
- `.cdf(x, ...)` - Función de probabilidad acumulada
- `.ppf(q, ...)` - Función cuantil (inversa de cdf)
- `.rvs(...)` - Generación de muestras aleatorias

**Ejemplo:**
```python
# Probabilidad acumulada de Z <= 1.96
stats.norm.cdf(1.96)

# Valor crítico para alpha = 0.05 bilateral
stats.norm.ppf(0.975)  # 1.96
```
