Una empresa va a lanzar un nuevo producto y ha realizado un estudio de mercado. Necesitan tomar decisiones basadas en múltiples hipótesis.

**Datos recopilados:**

```python
import pandas as pd
import numpy as np
from scipy import stats
import matplotlib.pyplot as plt
import seaborn as sns

# Configuración de visualización
sns.set_style("whitegrid")
plt.rcParams["figure.figsize"] = (10, 6)
```

```python
# Simulamos datos de un estudio de mercado
np.random.seed(789)

n_encuestados = 200

# Intención de compra (1 = sí, 0 = no)
intencion_compra = np.random.binomial(1, 0.35, n_encuestados)

# Precio dispuesto a pagar
precio_dispuesto = np.random.normal(50, 15, n_encuestados)
precio_dispuesto = np.clip(precio_dispuesto, 20, 100)  # limitar entre 20 y 100

# Segmento de edad
segmento_edad = np.random.choice(
    ["18-30", "31-45", "46-60"], n_encuestados, p=[0.4, 0.35, 0.25]
)

# Crear dataframe
data_producto = pd.DataFrame(
    {
        "Intencion_Compra": intencion_compra,
        "Precio_Dispuesto": precio_dispuesto,
        "Segmento_Edad": segmento_edad,
    }
)

print(data_producto.head(10))
print(f"\nResumen:")
print(
    f"Proporción con intención de compra: {data_producto['Intencion_Compra'].mean():.3f}"
)
print(
    f"Precio promedio dispuesto a pagar: ${data_producto['Precio_Dispuesto'].mean():.2f}"
)
```
