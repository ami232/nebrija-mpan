`scipy.stats` tiene todas las distribuciones con la misma interfaz:

- `stats.binom(n, p)` es una binomial y `stats.norm(mu, sigma)` una normal (¡`sigma` es la desviación típica, no la varianza!).
- `.pmf(k)`: $P(X = k)$ en una discreta.
- `.cdf(k)`: $P(X \le k)$.
- `.sf(k)`: $P(X > k) = 1 - F(k)$.
- `.ppf(q)`: el cuantil, el valor $x$ tal que $F(x) = q$.
- `.mean()`, `.std()`, `.var()`: media, desviación típica y varianza.
